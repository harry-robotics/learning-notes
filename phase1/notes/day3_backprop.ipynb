```python
import numpy as np

# sigmoid 激活函数：把任意实数压到 (0,1)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 一个样本的例子（先用标量，把机制看清楚）
x = 1.5      # 输入
y = 1.0      # 真实标签
w = 0.8      # 参数
b = 0.1      # 参数

# ---------- 前向传播：一步步算，并把中间量存下来 ----------
z = w * x + b            # 线性部分
a = sigmoid(z)           # 激活 = 预测值
L = (a - y) ** 2         # 平方损失
print(f"前向: z={z:.4f}, a={a:.4f}, L={L:.4f}")

# ---------- 反向传播：从 L 往回，每个节点 本地梯度 连乘 ----------
dL_da = 2 * (a - y)      # 本地: ∂L/∂a
da_dz = a * (1 - a)      # 本地: ∂a/∂z （sigmoid 导数）
dz_dw = x                # 本地: ∂z/∂w
dz_db = 1                # 本地: ∂z/∂b

dL_dw = dL_da * da_dz * dz_dw   # 链式法则连乘 → ∂L/∂w
dL_db = dL_da * da_dz * dz_db   # 链式法则连乘 → ∂L/∂b
print(f"反向传播梯度: dL/dw={dL_dw:.4f}, dL/db={dL_db:.4f}")

# ---------- 梯度检验：用"数值梯度"验证反向传播算对没 ----------
def loss_fn(w, b):
    return (sigmoid(w * x + b) - y) ** 2

eps = 1e-5
num_dw = (loss_fn(w + eps, b) - loss_fn(w - eps, b)) / (2 * eps)
num_db = (loss_fn(w, b + eps) - loss_fn(w, b - eps)) / (2 * eps)
print(f"数值梯度:     num_dw={num_dw:.4f}, num_db={num_db:.4f}")
```
