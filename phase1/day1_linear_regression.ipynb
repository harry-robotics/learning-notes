### 1.造一批模拟数据

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
n_samples = 100
X = np.random.rand(n_samples, 1) * 10
true_w , true_b = 2.0, 1.0
noise = np.random.randn(n_samples, 1) 
y = true_w * X + true_b + noise
```

### 2.模型：前向传播

```python
def forward(X, w, b):
    return X * w + b
```

### 3.损失：MSE

```python
def mse_loss(y_pred, y_true):
    return np.mean((y_pred - y_true) ** 2)
```

### 4.用一组随便猜的参数看效果

```python
w,b = 0.5,0.0
y_pred = forward(X, w, b)
loss = mse_loss(y_pred, y)
print(f"猜 w={w}, b={b}时的损失: {loss:.4f}")
```

### 5.画出数据加当前这条拟合线

```python
plt.scatter(X, y, s=10, label="data")
x_line = np.array([[0], [10]])
plt.plot(x_line, forward(x_line, w, b), color='red', label="current model")
plt.xlabel("x")
plt.ylabel("y"); plt.legend(); plt.show()
```

### 6.关键实验：看损失随w怎么变（固定 b=1）

```python
ws = np.linspace(-1, 5, 100)
losses = [mse_loss(forward(X, w_try, 1.0),y) for w_try in ws]
plt.plot(ws, losses)
plt.xlabel("w"); plt.ylabel("MSE loss"); plt.title("loss vs w")
plt.show()
```
